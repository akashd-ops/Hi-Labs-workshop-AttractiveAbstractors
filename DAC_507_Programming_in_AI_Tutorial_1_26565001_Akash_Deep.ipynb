```python
# Problem 3 solution : function that calculates price per sq inch given the dimaeter and price of pizza.
# medium size domino's pizza values taken!!
import math as mt
def pricepersqinch(price, diameter):
  return (4*price/ (mt.pi*diameter*diameter))

print(pricepersqinch(400, 10))


```

```python
# Problem 5 solution : Calculating Future investment value for 1lakh rupees at 8.5% interest rate for 10 years

def investmentvalue(principal, APR):
  t=10
  while(t):
    t = t - 1
    principal = principal*(1+APR)
  return principal
print(investmentvalue(100000, 0.085))
```

```python
# Problem 6 solution : Adding time variable for more general implementation

def investmentvalue(principal, APR,t):
  while(t):
    t = t - 1
    principal = principal*(1+APR)
  return principal
print(investmentvalue(100000, 0.085, 11))
```

```python
# Problem 7 Solution : Basically this is what a sip would look like if the stocks had constant ROI.
def investmentvalue(principal, annual_investment, APR, t):
    while t:
        principal = principal * (1 + APR)   # Add interest
        principal = principal + annual_investment  # Add yearly investment
        t -= 1
    return principal

# Example
print(investmentvalue(100000, 10000, 0.085, 11))
```

```python
#Problem 8 solution
def investmentvalue(principal, rate, periods, years):
    for year in range(years):
        for i in range(periods):
            principal = principal * (1 + rate / periods)
    return principal

#input
principal = float(input("Enter principal amount: "))
rate = float(input("Enter nominal annual interest rate (0.06 for 6%): "))
periods = int(input("Enter number of compounding periods per year: "))
years = int(input("Enter the time for which the investment is made in years: "))

print("Future Value =", investmentvalue(principal, rate, periods, years))
```
