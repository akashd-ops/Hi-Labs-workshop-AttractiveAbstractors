```python
# Problem 11 : factorial n calculation
def factorial(n):
  ans = 1
  for i in range(n):
    ans = ans *(i+1)
  return ans
print(factorial(5))
```

```python
# Problem 12: clearly when a list "numbers" is passed to the function for loop iterates over the list and calculate the sum of the elements in the list.
def total ( numbers ) :
 s = 0
 for n in numbers :
  s += n
 return s
print ( total ([1 , 2 , 3 , 4]) )
```

```python
# Problem 13
vowels = ['a', 'e', 'i', 'o', 'u']
def countvowels(word):
  count = 0
  for x in word:
    for y in vowels:
      if(x==y):
        count = count + 1
  return count
print(countvowels("standard_text"))
```

```python
# Problem 14 : Prints a pattern of n stars in a single n
def mystery (n):
 result = ""
 for i in range (n):
  result += "*"
 return result
print (mystery(4))
```

```python
# Problem 15
def reverse_string(s):
   m = len(s)
   reverse = ""
   for i in range(m):
    reverse = reverse + s[m-1-i]
   return reverse

print(reverse_string("Akash"))
```
